```python
# Memuat model 10000 step
import os
import sys
import tensorflow as tf

# Lokasi Object Detection API
RESEARCH_DIR = "/home/rzangef/SSD-PPE/object-detection-api/research"

if RESEARCH_DIR not in sys.path:
    sys.path.insert(0, RESEARCH_DIR)

from object_detection.utils import config_util
from object_detection.builders import model_builder

# Lokasi konfigurasi dan checkpoint
CONFIG_PATH = (
    "/home/rzangef/SSD-PPE/models/"
    "pipeline_ppe_yolo_split_continue_10000.config"
)

CHECKPOINT_PATH = (
    "/home/rzangef/SSD-PPE/training/"
    "ssd_mobilenet_v2_fpnlite_320x320_continue_10000/ckpt-5"
)

VAL_RECORD = (
    "/home/rzangef/SSD-PPE/dataset/"
    "tfrecord_yolo_split/val.record"
)

# Membaca konfigurasi
configs = config_util.get_configs_from_pipeline_file(CONFIG_PATH)
model_config = configs["model"]

# Membangun model untuk inferensi
detection_model = model_builder.build(
    model_config=model_config,
    is_training=False
)

# Memuat bobot checkpoint
checkpoint = tf.train.Checkpoint(model=detection_model)
status = checkpoint.restore(CHECKPOINT_PATH)

status.expect_partial()
status.assert_existing_objects_matched()

print("Model berhasil dimuat.")
print("Checkpoint:", CHECKPOINT_PATH)

# Membaca satu gambar validation
raw_dataset = tf.data.TFRecordDataset(VAL_RECORD)

for raw_record in raw_dataset.take(1):

    example = tf.train.Example()
    example.ParseFromString(raw_record.numpy())

    encoded_image = (
        example.features.feature["image/encoded"]
        .bytes_list.value[0]
    )

    image = tf.io.decode_image(
        encoded_image,
        channels=3,
        expand_animations=False
    )

    image = tf.cast(image, tf.float32)
    image = tf.expand_dims(image, axis=0)

    # Preprocessing
    preprocessed_image, true_image_shapes = (
        detection_model.preprocess(image)
    )

    # Prediksi
    prediction_dict = detection_model.predict(
        preprocessed_image,
        true_image_shapes
    )

    # Postprocessing
    detections = detection_model.postprocess(
        prediction_dict,
        true_image_shapes
    )

    print("\n===== HASIL INFERENSI SATU GAMBAR =====")
    print("Ukuran gambar:", image.shape)
    print(
        "Jumlah deteksi:",
        int(detections["num_detections"][0].numpy())
    )
    print(
        "Detection scores:",
        detections["detection_scores"][0][:10].numpy()
    )
    print(
        "Detection classes:",
        detections["detection_classes"][0][:10].numpy()
    )
```

```python
# Eksekusi Evaluasi Data Validasi

import tensorflow as tf
import numpy as np
import pandas as pd
from tqdm.auto import tqdm

# ==========================================
# KONFIGURASI EVALUASI
# ==========================================

VAL_RECORD = (
    "/home/rzangef/SSD-PPE/dataset/"
    "tfrecord_yolo_split/val.record"
)

CONF_THRESHOLD = 0.40
IOU_THRESHOLD = 0.50

# ==========================================
# FUNGSI MENGHITUNG IoU
# Format box: [xmin, ymin, xmax, ymax]
# ==========================================

def calculate_iou(box1, box2):

    x_left = max(box1[0], box2[0])
    y_top = max(box1[1], box2[1])

    x_right = min(box1[2], box2[2])
    y_bottom = min(box1[3], box2[3])

    intersection = (
        max(0, x_right - x_left) *
        max(0, y_bottom - y_top)
    )

    area1 = (
        max(0, box1[2] - box1[0]) *
        max(0, box1[3] - box1[1])
    )

    area2 = (
        max(0, box2[2] - box2[0]) *
        max(0, box2[3] - box2[1])
    )

    union = area1 + area2 - intersection

    return intersection / union if union > 0 else 0.0


# ==========================================
# INISIALISASI COUNTER
# ==========================================

TP = 0
FP = 0
FN = 0
ignored_predictions = 0

total_images = 0
total_ground_truth = 0

dataset = tf.data.TFRecordDataset(VAL_RECORD)

# ==========================================
# EVALUASI SETIAP GAMBAR
# ==========================================

for raw_record in tqdm(dataset, desc="Evaluasi Validation"):

    example = tf.train.Example()
    example.ParseFromString(raw_record.numpy())

    features = example.features.feature

    # Decode gambar
    encoded_image = (
        features["image/encoded"].bytes_list.value[0]
    )

    image = tf.io.decode_image(
        encoded_image,
        channels=3,
        expand_animations=False
    )

    image = tf.cast(image, tf.float32)
    image = image[tf.newaxis, ...]

    # ======================================
    # GROUND TRUTH
    # ======================================

    gt_classes = np.array(
        features["image/object/class/label"]
        .int64_list.value,
        dtype=np.int32
    )

    xmin = np.array(
        features["image/object/bbox/xmin"]
        .float_list.value
    )

    ymin = np.array(
        features["image/object/bbox/ymin"]
        .float_list.value
    )

    xmax = np.array(
        features["image/object/bbox/xmax"]
        .float_list.value
    )

    ymax = np.array(
        features["image/object/bbox/ymax"]
        .float_list.value
    )

    difficult = np.array(
        features["image/object/difficult"]
        .int64_list.value
        if "image/object/difficult" in features
        else [0] * len(gt_classes),
        dtype=np.int32
    )

    # Ground truth dalam format XYXY
    gt_boxes = np.column_stack([
        xmin, ymin, xmax, ymax
    ]).astype(np.float32)

    valid_gt = difficult == 0
    difficult_gt = difficult == 1

    gt_boxes_valid = gt_boxes[valid_gt]
    gt_classes_valid = gt_classes[valid_gt]

    gt_boxes_difficult = gt_boxes[difficult_gt]
    gt_classes_difficult = gt_classes[difficult_gt]

    total_ground_truth += len(gt_classes_valid)
    total_images += 1

    # ======================================
    # INFERENSI MODEL
    # ======================================

    preprocessed_image, true_image_shapes = (
        detection_model.preprocess(image)
    )

    prediction_dict = detection_model.predict(
        preprocessed_image,
        true_image_shapes
    )

    detections = detection_model.postprocess(
        prediction_dict,
        true_image_shapes
    )

    num_det = int(
        detections["num_detections"][0].numpy()
    )

    pred_boxes_raw = (
        detections["detection_boxes"][0][:num_det].numpy()
    )

    pred_scores = (
        detections["detection_scores"][0][:num_det].numpy()
    )

    pred_classes_raw = (
        detections["detection_classes"][0][:num_det].numpy()
    )

    # Filter confidence
    selected = pred_scores >= CONF_THRESHOLD

    pred_boxes_raw = pred_boxes_raw[selected]
    pred_scores = pred_scores[selected]
    pred_classes_raw = pred_classes_raw[selected]

    # Model menghasilkan class ID 0-7.
    # Dataset menggunakan class ID 1-8.
    pred_classes = (
        pred_classes_raw.astype(np.int32) + 1
    )

    # Konversi box model [ymin,xmin,ymax,xmax]
    # menjadi [xmin,ymin,xmax,ymax]
    pred_boxes = np.column_stack([
        pred_boxes_raw[:, 1],
        pred_boxes_raw[:, 0],
        pred_boxes_raw[:, 3],
        pred_boxes_raw[:, 2]
    ]).astype(np.float32)

    # ======================================
    # MATCHING PREDIKSI DAN GROUND TRUTH
    # ======================================

    matched_gt = set()

    # Urutkan prediksi dari confidence tertinggi
    order = np.argsort(-pred_scores)

    for pred_idx in order:

        pred_box = pred_boxes[pred_idx]
        pred_class = pred_classes[pred_idx]

        candidate_indices = np.where(
            gt_classes_valid == pred_class
        )[0]

        best_iou = 0.0
        best_gt_idx = None

        # Cari ground truth dengan IoU tertinggi
        # yang belum pernah dipasangkan.
        for gt_idx in candidate_indices:

            if gt_idx in matched_gt:
                continue

            iou = calculate_iou(
                pred_box,
                gt_boxes_valid[gt_idx]
            )

            if iou > best_iou:
                best_iou = iou
                best_gt_idx = gt_idx

        if (
            best_gt_idx is not None
            and best_iou >= IOU_THRESHOLD
        ):
            TP += 1
            matched_gt.add(best_gt_idx)

        else:

            # Prediksi yang cocok dengan objek difficult
            # tidak dihitung sebagai TP maupun FP.
            difficult_indices = np.where(
                gt_classes_difficult == pred_class
            )[0]

            overlaps_difficult = any(
                calculate_iou(
                    pred_box,
                    gt_boxes_difficult[idx]
                ) >= IOU_THRESHOLD
                for idx in difficult_indices
            )

            if overlaps_difficult:
                ignored_predictions += 1
            else:
                FP += 1

    # FN: GT valid yang tidak mendapatkan pasangan
    FN += len(gt_classes_valid) - len(matched_gt)

    if total_images % 20 == 0:
        print(
            f"\nGambar: {total_images}/170 | "
            f"TP: {TP} | FP: {FP} | FN: {FN}"
        )


# ==========================================
# HASIL AKHIR
# ==========================================

precision = TP / (TP + FP) if (TP + FP) else 0.0

recall = TP / (TP + FN) if (TP + FN) else 0.0

f1_score = (
    2 * precision * recall / (precision + recall)
    if (precision + recall) else 0.0
)

print("\n========================================")
print("HASIL EVALUASI SSD MOBILENET")
print("VALIDATION SET - 10.000 STEP")
print("========================================")

print("Jumlah gambar :", total_images)
print("Ground truth  :", total_ground_truth)
print("TP            :", TP)
print("FP            :", FP)
print("FN            :", FN)
print("Ignored       :", ignored_predictions)

print("----------------------------------------")
print(f"Precision     : {precision:.4f} ({precision*100:.2f}%)")
print(f"Recall        : {recall:.4f} ({recall*100:.2f}%)")
print(f"F1-score      : {f1_score:.4f} ({f1_score*100:.2f}%)")
print("========================================")
```
