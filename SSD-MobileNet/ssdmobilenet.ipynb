```python
# Ubah file freezable_sync_batch_norm.py
path = "/home/rzangef/spark-data/models/research/object_detection/core/freezable_sync_batch_norm.py"

with open(path, "r") as f:
    content = f.read()

old = "tf.keras.layers.experimental.SyncBatchNormalization"
new = "tf.keras.layers.BatchNormalization"

if old in content:
    content = content.replace(old, new)

    with open(path, "w") as f:
        f.write(content)

    print("Patch berhasil diterapkan.")
else:
    print("Teks target tidak ditemukan.")
```
