```python
from pathlib import Path
import xml.etree.ElementTree as ET
from collections import Counter

ROOT = Path(r"C:\Faster-RCNN-PPE\dataset\images")

class_counts = Counter()

total_images = 0
total_objects = 0
empty_xml = []
invalid_xml = []
invalid_boxes = []
outside_boxes = []

for split in ["train", "val", "test"]:
    folder = ROOT / split

    for xml_file in folder.glob("*.xml"):
        total_images += 1

        try:
            tree = ET.parse(xml_file)
            root = tree.getroot()
        except Exception as e:
            invalid_xml.append((split, xml_file.name, str(e)))
            continue

        size = root.find("size")

        if size is None:
            invalid_xml.append((split, xml_file.name, "Tidak ada elemen size"))
            continue

        width = int(size.findtext("width", "0"))
        height = int(size.findtext("height", "0"))

        objects = root.findall("object")

        if not objects:
            empty_xml.append((split, xml_file.name))

        for obj in objects:
            name = obj.findtext("name")

            if not name:
                invalid_boxes.append((split, xml_file.name, "Nama kelas kosong"))
                continue

            class_counts[name] += 1
            total_objects += 1

            bbox = obj.find("bndbox")

            if bbox is None:
                invalid_boxes.append((split, xml_file.name, "Bounding box tidak ditemukan"))
                continue

            try:
                xmin = float(bbox.findtext("xmin"))
                ymin = float(bbox.findtext("ymin"))
                xmax = float(bbox.findtext("xmax"))
                ymax = float(bbox.findtext("ymax"))
            except (TypeError, ValueError):
                invalid_boxes.append((split, xml_file.name, "Koordinat tidak valid"))
                continue

            if xmin >= xmax or ymin >= ymax:
                invalid_boxes.append((split, xml_file.name, "Ukuran bounding box tidak valid"))

            if xmin < 0 or ymin < 0 or xmax > width or ymax > height:
                outside_boxes.append((split, xml_file.name))

print("===== HASIL AUDIT DATASET =====")
print("Total XML diperiksa:", total_images)
print("Total objek:", total_objects)

print("\n===== JUMLAH OBJEK PER KELAS =====")
for name, count in sorted(class_counts.items()):
    print(f"{name}: {count}")

print("\n===== PEMERIKSAAN ANOTASI =====")
print("XML tanpa objek:", len(empty_xml))
print("XML rusak/tidak lengkap:", len(invalid_xml))
print("Bounding box tidak valid:", len(invalid_boxes))
print("Bounding box di luar gambar:", len(outside_boxes))

if empty_xml:
    print("\nContoh XML tanpa objek:", empty_xml[:5])

if invalid_xml:
    print("\nContoh XML bermasalah:", invalid_xml[:5])

if invalid_boxes:
    print("\nContoh bounding box tidak valid:", invalid_boxes[:5])

if outside_boxes:
    print("\nContoh bounding box di luar gambar:", outside_boxes[:5])
```

```python
# Membuat Dataset class Faster R-CNN
from pathlib import Path
import xml.etree.ElementTree as ET

import torch
from torch.utils.data import Dataset
from PIL import Image
from torchvision.transforms.functional import to_tensor

# Lokasi dataset
ROOT = Path(r"C:\Faster-RCNN-PPE\dataset\images")

# Pemetaan kelas XML ke ID Faster R-CNN
CLASS_MAP = {
    "person": 1,
    "vest": 2,
    "yellow": 3,
    "white": 4,
    "glass": 5,
    "blue": 6,
    "red": 7,
    "head": 8
}

class CHVGPpeDataset(Dataset):

    def __init__(self, root, split):
        self.folder = Path(root) / split

        self.images = sorted([
            f for f in self.folder.iterdir()
            if f.is_file() and f.suffix.lower() in [".jpg", ".jpeg"]
        ])

    def __len__(self):
        return len(self.images)

    def __getitem__(self, idx):

        image_path = self.images[idx]
        xml_path = self.folder / (image_path.stem + ".xml")

        # Baca gambar
        image = Image.open(image_path).convert("RGB")
        width, height = image.size

        # Baca XML
        root = ET.parse(xml_path).getroot()

        boxes = []
        labels = []

        for obj in root.findall("object"):

            class_name = obj.findtext("name")

            if class_name not in CLASS_MAP:
                raise ValueError(
                    f"Kelas tidak dikenal: {class_name}"
                )

            bbox = obj.find("bndbox")

            xmin = float(bbox.findtext("xmin"))
            ymin = float(bbox.findtext("ymin"))
            xmax = float(bbox.findtext("xmax"))
            ymax = float(bbox.findtext("ymax"))

            # Clipping koordinat ke batas gambar
            xmin = max(0, min(xmin, width))
            ymin = max(0, min(ymin, height))
            xmax = max(0, min(xmax, width))
            ymax = max(0, min(ymax, height))

            # Abaikan box yang menjadi tidak valid setelah clipping
            if xmin >= xmax or ymin >= ymax:
                continue

            boxes.append([xmin, ymin, xmax, ymax])
            labels.append(CLASS_MAP[class_name])

        # Konversi gambar menjadi tensor
        image_tensor = to_tensor(image)

        # Tensor anotasi
        boxes_tensor = torch.tensor(
            boxes, dtype=torch.float32
        ).reshape(-1, 4)

        labels_tensor = torch.tensor(
            labels, dtype=torch.int64
        )

        target = {
            "boxes": boxes_tensor,
            "labels": labels_tensor,
            "image_id": torch.tensor([idx]),
            "area": (
                (boxes_tensor[:, 2] - boxes_tensor[:, 0]) *
                (boxes_tensor[:, 3] - boxes_tensor[:, 1])
            ),
            "iscrowd": torch.zeros(
                (len(labels),), dtype=torch.int64
            )
        }

        return image_tensor, target


# Membuat dataset untuk setiap split
train_dataset = CHVGPpeDataset(ROOT, "train")
val_dataset = CHVGPpeDataset(ROOT, "val")
test_dataset = CHVGPpeDataset(ROOT, "test")

print("Dataset berhasil dibuat")
print("Train:", len(train_dataset))
print("Validation:", len(val_dataset))
print("Test:", len(test_dataset))
```

```python
# Validasi seluruh anotasi hasil dataset class
from collections import Counter

datasets = {
    "train": train_dataset,
    "val": val_dataset,
    "test": test_dataset
}

for split, dataset in datasets.items():

    class_counts = Counter()
    total_objects = 0
    empty_images = 0

    for i in range(len(dataset)):
        image, target = dataset[i]

        labels = target["labels"].tolist()

        if len(labels) == 0:
            empty_images += 1

        class_counts.update(labels)
        total_objects += len(labels)

    print(f"\n===== {split.upper()} =====")
    print("Jumlah gambar:", len(dataset))
    print("Jumlah objek:", total_objects)
    print("Gambar tanpa objek:", empty_images)

    print("Distribusi kelas:")
    for class_id in range(1, 9):
        print(f"{class_id} - {list(CLASS_MAP.keys())[class_id-1]}: {class_counts[class_id]}")
```

```python
# Membuat Data Loader
from torch.utils.data import DataLoader

def collate_fn(batch):
    return tuple(zip(*batch))

train_loader = DataLoader(
    train_dataset,
    batch_size=1,
    shuffle=True,
    num_workers=0,
    collate_fn=collate_fn
)

val_loader = DataLoader(
    val_dataset,
    batch_size=1,
    shuffle=False,
    num_workers=0,
    collate_fn=collate_fn
)

print("DataLoader berhasil dibuat")
print("Jumlah batch train:", len(train_loader))
print("Jumlah batch validation:", len(val_loader))
```

```python
# Membuat model
import torch
import torch_directml
from torchvision.models.detection import fasterrcnn_resnet50_fpn
from torchvision.models.detection.faster_rcnn import FastRCNNPredictor

# 1. Menentukan perangkat DirectML
device = torch_directml.device()

print("Device:", device)
print("Nama GPU:", torch_directml.device_name(0))

# 2. Membuat model Faster R-CNN
model = fasterrcnn_resnet50_fpn(
    weights=None,
    weights_backbone=None,
    num_classes=9,
    min_size=640,
    max_size=640
)

model.to(device)
model.train()

print("Model berhasil dibuat dan dipindahkan ke DirectML")
```

```python
# Uji training 1 batch
import torch

# 1. Ambil satu batch data training
images, targets = next(iter(train_loader))

# 2. Pindahkan gambar dan anotasi ke DirectML
images = [img.to(device) for img in images]

targets = [
    {
        key: value.to(device) if torch.is_tensor(value) else value
        for key, value in target.items()
    }
    for target in targets
]

# 3. Siapkan optimizer
optimizer = torch.optim.SGD(
    model.parameters(),
    lr=0.001,
    momentum=0.9,
    weight_decay=0.0005
)

# 4. Forward: menghitung loss
model.train()
loss_dict = model(images, targets)

total_loss = sum(loss for loss in loss_dict.values())

print("===== HASIL FORWARD =====")
for name, loss in loss_dict.items():
    print(f"{name}: {loss.item():.6f}")

print("Total loss:", total_loss.item())

# 5. Backward: menghitung gradien
optimizer.zero_grad()
total_loss.backward()

print("\nBackward berhasil")

# 6. Optimizer step
optimizer.step()

print("Optimizer step berhasil")
```

```python
# Mengunduh petrained COCO
import torch
import torch_directml

from torchvision.models.detection import (
    fasterrcnn_resnet50_fpn,
    FasterRCNN_ResNet50_FPN_Weights
)

from torchvision.models.detection.faster_rcnn import FastRCNNPredictor

# 1. Memuat bobot pretrained COCO
weights = FasterRCNN_ResNet50_FPN_Weights.DEFAULT

model = fasterrcnn_resnet50_fpn(
    weights=weights,
    min_size=640,
    max_size=640
)

print("Pretrained COCO berhasil dimuat")

# 2. Mengganti classifier untuk 8 kelas APD + background
in_features = model.roi_heads.box_predictor.cls_score.in_features

model.roi_heads.box_predictor = FastRCNNPredictor(
    in_features,
    num_classes=9
)

print("Classifier berhasil disesuaikan")
print("Jumlah kelas:", 9)

# 3. Memindahkan model ke DirectML
device = torch_directml.device()
model.to(device)

print("Device:", device)
print("GPU:", torch_directml.device_name(0))

# 4. Memastikan mode training
model.train()

print("Model Faster R-CNN siap untuk pengujian berikutnya")
```

```python
# Uji training 5 batch
import time
import torch

# Pastikan model dalam mode training
model.train()

# Optimizer
optimizer = torch.optim.SGD(
    model.parameters(),
    lr=0.001,
    momentum=0.9,
    weight_decay=0.0005
)

# Ambil 5 batch pertama
small_loader = iter(train_loader)

batch_times = []
loss_values = []

print("===== UJI TRAINING 5 BATCH =====")

for step in range(5):
    start_time = time.time()

    images, targets = next(small_loader)

    # Pindahkan data ke DirectML
    images = [img.to(device) for img in images]

    targets = [
        {
            key: value.to(device) if torch.is_tensor(value) else value
            for key, value in target.items()
        }
        for target in targets
    ]

    # Forward
    loss_dict = model(images, targets)
    total_loss = sum(loss for loss in loss_dict.values())

    # Backward dan update bobot
    optimizer.zero_grad()
    total_loss.backward()
    optimizer.step()

    elapsed = time.time() - start_time

    batch_times.append(elapsed)
    loss_values.append(total_loss.item())

    print(
        f"Batch {step+1}/5 | "
        f"Loss: {total_loss.item():.4f} | "
        f"Waktu: {elapsed:.2f} detik"
    )

print("\n===== RINGKASAN =====")
print("Rata-rata waktu per batch:",
      round(sum(batch_times) / len(batch_times), 2), "detik")

print("Total waktu:",
      round(sum(batch_times), 2), "detik")

print("Rata-rata loss:",
      round(sum(loss_values) / len(loss_values), 4))
```

```python
# Training 1 epoch

import os
import time
import torch
import torch_directml

# ==========================================
# 1. KONFIGURASI DEVICE DAN TRAINING
# ==========================================

device = torch_directml.device()

# Pastikan model berada di DirectML
model.to(device)
model.train()

# Optimizer SGD
optimizer = torch.optim.SGD(
    model.parameters(),
    lr=0.001,
    momentum=0.9,
    weight_decay=0.0005
)

print("Device:", torch_directml.device_name(0))
print("Jumlah gambar training:", len(train_loader))
print("Batch size: 1")
print("Epoch: 1")
print("Learning rate: 0.001")

# ==========================================
# 2. FUNGSI UNTUK MENYIMPAN CHECKPOINT
# ==========================================

def move_to_cpu(data):
    if isinstance(data, torch.Tensor):
        return data.detach().cpu()
    elif isinstance(data, dict):
        return {k: move_to_cpu(v) for k, v in data.items()}
    elif isinstance(data, list):
        return [move_to_cpu(v) for v in data]
    elif isinstance(data, tuple):
        return tuple(move_to_cpu(v) for v in data)
    return data

# ==========================================
# 3. PROSES TRAINING 1 EPOCH
# ==========================================

total_loss = 0.0
start_time = time.time()

for batch_idx, (images, targets) in enumerate(train_loader):

    # Memindahkan gambar dan anotasi ke DirectML
    images = [img.to(device) for img in images]

    targets = [
        {
            key: value.to(device) if isinstance(value, torch.Tensor) else value
            for key, value in target.items()
        }
        for target in targets
    ]

    # Menghapus gradien sebelumnya
    optimizer.zero_grad()

    # Forward pass dan menghitung loss
    loss_dict = model(images, targets)
    losses = sum(loss for loss in loss_dict.values())

    # Memastikan loss valid
    if not torch.isfinite(losses).item():
        raise RuntimeError(
            f"Loss tidak valid pada batch {batch_idx + 1}: "
            f"{losses.item()}"
        )

    # Backward pass dan pembaruan bobot
    losses.backward()
    optimizer.step()

    batch_loss = losses.item()
    total_loss += batch_loss

    # Menampilkan progres setiap 100 batch
    if (batch_idx + 1) % 100 == 0 or (batch_idx + 1) == len(train_loader):
        print(
            f"Epoch 1 | Batch {batch_idx + 1}/{len(train_loader)} "
            f"| Loss: {batch_loss:.4f}"
        )

# ==========================================
# 4. HASIL TRAINING
# ==========================================

avg_loss = total_loss / len(train_loader)
elapsed_time = time.time() - start_time

print("\n========== HASIL TRAINING ==========")
print("Epoch              : 1")
print("Jumlah batch       :", len(train_loader))
print(f"Total loss         : {total_loss:.4f}")
print(f"Average loss       : {avg_loss:.4f}")
print(f"Durasi             : {elapsed_time / 60:.2f} menit")
print("====================================")

# ==========================================
# 5. SIMPAN CHECKPOINT
# ==========================================

checkpoint_dir = r"C:\Faster-RCNN-PPE\checkpoints"
os.makedirs(checkpoint_dir, exist_ok=True)

checkpoint_path = os.path.join(
    checkpoint_dir,
    "fasterrcnn_epoch_1.pth"
)

checkpoint = {
    "epoch": 1,
    "model_state_dict": move_to_cpu(model.state_dict()),
    "optimizer_state_dict": move_to_cpu(optimizer.state_dict()),
    "average_loss": avg_loss,
    "learning_rate": 0.001
}

torch.save(checkpoint, checkpoint_path)

print("\nCheckpoint berhasil disimpan:")
print(checkpoint_path)
```
