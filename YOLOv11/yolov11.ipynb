```python
# Buat folder dataset YOLO
import os
import glob
import xml.etree.ElementTree as ET

source_path = "/home/rzangef/spark-data/CHVG-Dataset-original"
yolo_path = "/home/rzangef/spark-data/CHVG-YOLO"

folders = [
    os.path.join(yolo_path, "images", "train"),
    os.path.join(yolo_path, "images", "val"),
    os.path.join(yolo_path, "images", "test"),
    os.path.join(yolo_path, "labels", "train"),
    os.path.join(yolo_path, "labels", "val"),
    os.path.join(yolo_path, "labels", "test")
]

for folder in folders:
    os.makedirs(folder, exist_ok=True)

print("Folder dataset YOLO berhasil dibuat.")

# Cek
print(os.listdir(yolo_path))
```

```python
# Cari seluruh file XML
source_path = "/home/rzangef/spark-data/CHVG-Dataset-original"

xml_files = glob.glob(
    os.path.join(source_path, "**", "*.xml"),
    recursive=True
)

print("Jumlah XML ditemukan:", len(xml_files))
print("Contoh XML:", xml_files[0])
```

```python
# Cari seluruh file XML
xml_files = glob.glob(
    os.path.join(source_path, "**", "*.xml"),
    recursive=True
)

print("Jumlah XML ditemukan:", len(xml_files))
```

```python
# Buat folder untuk menyimpan TXT
label_output = os.path.join(
    yolo_path,
    "labels",
    "all"
)

os.makedirs(label_output, exist_ok=True)
```

```python
# Menentukan Class ID

classes = [
    "person",
    "vest",
    "yellow",
    "white",
    "glass",
    "blue",
    "red",
    "head"
]

class_to_id = {name: idx for idx, name in enumerate(classes)}

print("Class ID:")
for class_name, class_id in class_to_id.items():
    print(f"{class_id}: {class_name}")
```

```python
# Konversi XML ke TXT

converted = 0
failed = 0

for xml_file in xml_files:

    try:
        tree = ET.parse(xml_file)
        root = tree.getroot()

        width = int(root.find("size/width").text)
        height = int(root.find("size/height").text)

        filename = os.path.splitext(
            os.path.basename(xml_file)
        )[0]

        txt_file = os.path.join(
            label_output,
            filename + ".txt"
        )

        yolo_lines = []

        for obj in root.findall("object"):

            class_name = obj.find("name").text.strip()

            if class_name not in class_to_id:
                print(
                    f"Class tidak dikenal: {class_name} "
                    f"pada file {xml_file}"
                )
                continue

            class_id = class_to_id[class_name]

            bbox = obj.find("bndbox")

            xmin = float(bbox.find("xmin").text)
            ymin = float(bbox.find("ymin").text)
            xmax = float(bbox.find("xmax").text)
            ymax = float(bbox.find("ymax").text)

            center_x = ((xmin + xmax) / 2) / width
            center_y = ((ymin + ymax) / 2) / height

            bbox_width = (xmax - xmin) / width
            bbox_height = (ymax - ymin) / height

            yolo_line = (
                f"{class_id} "
                f"{center_x:.6f} "
                f"{center_y:.6f} "
                f"{bbox_width:.6f} "
                f"{bbox_height:.6f}"
            )

            yolo_lines.append(yolo_line)

        with open(txt_file, "w") as f:
            f.write("\n".join(yolo_lines))

        converted += 1

    except Exception as e:
        print(f"Gagal memproses: {xml_file}")
        print(f"Error: {e}")
        failed += 1

print("\n====================================")
print("KONVERSI SELESAI")
print("====================================")
print("Jumlah XML :", len(xml_files))
print("Berhasil   :", converted)
print("Gagal      :", failed)
```

```python
# Validasi Label YOLO
import os
import glob

yolo_path = "/home/rzangef/spark-data/CHVG-YOLO"

# Folder label
label_dirs = [
    "train",
    "val",
    "test"
]

total_labels = 0
total_objects = 0
invalid_format = 0
invalid_class = 0
invalid_coordinate = 0
empty_labels = 0

for split in label_dirs:

    label_path = os.path.join(
        yolo_path,
        "labels",
        split
    )

    txt_files = glob.glob(
        os.path.join(label_path, "*.txt")
    )

    print(f"\nMemeriksa {split.upper()}...")
    print("Jumlah TXT:", len(txt_files))

    for txt_file in txt_files:

        total_labels += 1

        with open(txt_file, "r") as f:
            lines = f.readlines()

        # Cek label kosong
        if len(lines) == 0:
            empty_labels += 1
            continue

        for line in lines:

            parts = line.strip().split()

            # Format harus 5 nilai
            if len(parts) != 5:
                invalid_format += 1
                continue

            try:
                class_id = int(parts[0])

                x_center = float(parts[1])
                y_center = float(parts[2])
                width = float(parts[3])
                height = float(parts[4])

            except ValueError:
                invalid_format += 1
                continue

            total_objects += 1

            # Cek class ID
            if class_id < 0 or class_id > 7:
                invalid_class += 1

            # Cek koordinat
            values = [
                x_center,
                y_center,
                width,
                height
            ]

            if any(value < 0 or value > 1 for value in values):
                invalid_coordinate += 1


print("\n====================================")
print("HASIL VALIDASI LABEL YOLO")
print("====================================")

print("Total file TXT       :", total_labels)
print("Total objek          :", total_objects)
print("Format tidak valid  :", invalid_format)
print("Class ID tidak valid:", invalid_class)
print("Koordinat tidak valid:", invalid_coordinate)
print("Label kosong         :", empty_labels)
```

```python
# Ubah Split Dataset
import os
import glob
import random
import shutil

# ==========================================
# PATH DATASET
# ==========================================

source_path = "/home/rzangef/spark-data/CHVG-Dataset-original"
yolo_path = "/home/rzangef/spark-data/CHVG-YOLO"

image_source = os.path.join(
    source_path,
    "CHVG-Dataset"
)

label_source = os.path.join(
    yolo_path,
    "labels",
    "all"
)

# ==========================================
# AMBIL SEMUA JPG
# ==========================================

jpg_files = glob.glob(
    os.path.join(image_source, "*.jpg")
)

print("Jumlah JPG:", len(jpg_files))

# ==========================================
# ACAK DATA DENGAN SEED TETAP
# ==========================================

random.seed(42)
random.shuffle(jpg_files)

# ==========================================
# JUMLAH DATA
# ==========================================

train_count = 1189
val_count = 170
test_count = 340

print("\nPembagian dataset:")
print("Train      :", train_count)
print("Validation :", val_count)
print("Test       :", test_count)
print("Total      :", train_count + val_count + test_count)

assert len(jpg_files) == 1699
assert train_count + val_count + test_count == 1699

# ==========================================
# BAGI DATA
# ==========================================

train_files = jpg_files[:train_count]

val_files = jpg_files[
    train_count:train_count + val_count
]

test_files = jpg_files[
    train_count + val_count:
]

# ==========================================
# HAPUS ISI SPLIT LAMA
# ==========================================

for split in ["train", "val", "test"]:

    image_dir = os.path.join(
        yolo_path,
        "images",
        split
    )

    label_dir = os.path.join(
        yolo_path,
        "labels",
        split
    )

    for file in glob.glob(
        os.path.join(image_dir, "*")
    ):
        os.remove(file)

    for file in glob.glob(
        os.path.join(label_dir, "*")
    ):
        os.remove(file)

# ==========================================
# COPY DATA BARU
# ==========================================

splits = {
    "train": train_files,
    "val": val_files,
    "test": test_files
}

for split, files in splits.items():

    image_output = os.path.join(
        yolo_path,
        "images",
        split
    )

    label_output = os.path.join(
        yolo_path,
        "labels",
        split
    )

    for jpg_file in files:

        filename = os.path.splitext(
            os.path.basename(jpg_file)
        )[0]

        txt_file = os.path.join(
            label_source,
            filename + ".txt"
        )

        # Copy JPG
        shutil.copy2(
            jpg_file,
            os.path.join(
                image_output,
                os.path.basename(jpg_file)
            )
        )

        # Copy TXT
        if os.path.exists(txt_file):

            shutil.copy2(
                txt_file,
                os.path.join(
                    label_output,
                    filename + ".txt"
                )
            )

        else:
            print(
                "WARNING: TXT tidak ditemukan:",
                filename
            )

print("\n====================================")
print("SPLIT 70/10/20 SELESAI")
print("====================================")
```
