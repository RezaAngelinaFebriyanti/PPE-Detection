```python
# smoke test inference pada 1 gambar CHVG menggunakan YOLOv11s + DirectML
import torch
import torch_directml
from PIL import Image

# Ambil satu gambar test
image_path = image_files[0]

# Pastikan model berada di DirectML
model.model.to(dml)

# Buka gambar
img = Image.open(image_path).convert("RGB")
img = img.resize((640, 640))

# Konversi PIL Image langsung ke tensor, tanpa NumPy
img_bytes = img.tobytes()
img_tensor = torch.tensor(
    list(img_bytes),
    dtype=torch.uint8
).reshape(640, 640, 3)

# Ubah format HWC -> CHW dan normalisasi
img_tensor = img_tensor.permute(2, 0, 1).float() / 255.0
img_tensor = img_tensor.unsqueeze(0).to(dml)

print("Image :", image_path)
print("Input :", img_tensor.shape)
print("Device:", img_tensor.device)

with torch.no_grad():
    output = model.model(img_tensor)

print("YOLO inference forward: BERHASIL")
print("Output type:", type(output))
```

```python
# Ambil kandidat prediksi
# Ambil output deteksi
pred = output[0]

# Pindahkan ke CPU untuk post-processing
pred_cpu = pred.detach().cpu()

# Hilangkan dimensi batch
pred_cpu = pred_cpu[0]

# Pisahkan koordinat box dan skor kelas
boxes = pred_cpu[:4, :]       # 4 koordinat bounding box
class_scores = pred_cpu[4:, :]  # 8 skor kelas

# Cari kelas dan skor tertinggi untuk setiap kandidat
conf, class_ids = class_scores.max(dim=0)

print("Jumlah kandidat :", pred_cpu.shape[1])
print("Jumlah kelas    :", class_scores.shape[0])
print("Confidence max  :", conf.max().item())
print("Confidence mean :", conf.mean().item())

# Kandidat dengan confidence >= 0.25
threshold = 0.25
mask = conf >= threshold

print("Kandidat >= 0.25:", mask.sum().item())
```

```python
# Prediksi top 10 confidence tertinggi
top_k = 10

top_conf, top_idx = torch.topk(conf, k=top_k)

print("10 prediksi dengan confidence tertinggi:\n")

class_names = [
    "person",
    "vest",
    "yellow",
    "white",
    "glass",
    "blue",
    "red",
    "head"
]

for rank in range(top_k):
    idx = top_idx[rank].item()
    confidence = top_conf[rank].item()
    class_id = class_ids[idx].item()

    box = boxes[:, idx]

    print(
        f"{rank+1:2d}. "
        f"class={class_id} ({class_names[class_id]}), "
        f"confidence={confidence:.4f}, "
        f"box=[{box[0]:.1f}, {box[1]:.1f}, "
        f"{box[2]:.1f}, {box[3]:.1f}]"
    )
```

```python
# Forward Pass
import torch
import torch_directml
from PIL import Image

# DirectML
dml = torch_directml.device()

# Gunakan gambar test pertama
image_path = image_files[0]

# Baca dan resize gambar
img = Image.open(image_path).convert("RGB")
img = img.resize((640, 640))

# Konversi gambar tanpa NumPy
img_bytes = img.tobytes()

img_tensor = torch.tensor(
    list(img_bytes),
    dtype=torch.uint8
).reshape(640, 640, 3)

img_tensor = img_tensor.permute(2, 0, 1).float() / 255.0
img_tensor = img_tensor.unsqueeze(0).to(dml)

# Pindahkan model ke DirectML
model.model.to(dml)

# Forward pass
with torch.no_grad():
    output = model.model(img_tensor)

# Ambil prediction tensor
pred = output[0].detach().cpu()[0]

# Pisahkan bounding box dan skor kelas
boxes = pred[:4, :]
class_scores = pred[4:, :]

# Ambil confidence dan class tertinggi
conf, class_ids = class_scores.max(dim=0)

print("Forward pass        : BERHASIL")
print("Jumlah kandidat     :", pred.shape[1])
print("Jumlah kelas        :", class_scores.shape[0])
print("Confidence tertinggi:", conf.max().item())
print("Kandidat >= 0.25    :", (conf >= 0.25).sum().item())
```

```python
# Prediksi top 10 confidence tertinggi
top_k = 10

top_conf, top_idx = torch.topk(conf, k=top_k)

print("10 prediksi dengan confidence tertinggi:\n")

class_names = [
    "person",
    "vest",
    "yellow",
    "white",
    "glass",
    "blue",
    "red",
    "head"
]

for rank in range(top_k):
    idx = top_idx[rank].item()
    confidence = top_conf[rank].item()
    class_id = class_ids[idx].item()

    box = boxes[:, idx]

    print(
        f"{rank+1:2d}. "
        f"class={class_id} ({class_names[class_id]}), "
        f"confidence={confidence:.4f}, "
        f"box=[{box[0]:.1f}, {box[1]:.1f}, "
        f"{box[2]:.1f}, {box[3]:.1f}]"
    )
```

```python
# NMS pada 28 kandidat dengan confidence ≥ 0.25
import torchvision

# Ambil kandidat dengan confidence >= 0.25
threshold = 0.25
mask = conf >= threshold

boxes_selected = boxes[:, mask]
scores_selected = conf[mask]
classes_selected = class_ids[mask]

# Konversi xywh -> xyxy
x_center = boxes_selected[0]
y_center = boxes_selected[1]
width = boxes_selected[2]
height = boxes_selected[3]

x1 = x_center - width / 2
y1 = y_center - height / 2
x2 = x_center + width / 2
y2 = y_center + height / 2

boxes_xyxy = torch.stack([x1, y1, x2, y2], dim=1)

# NMS per kelas
keep = torchvision.ops.batched_nms(
    boxes_xyxy,
    scores_selected,
    classes_selected,
    iou_threshold=0.45
)

print("Kandidat sebelum NMS :", len(scores_selected))
print("Deteksi setelah NMS  :", len(keep))

print("\nDeteksi final:")
for i in keep:
    idx = i.item()

    print(
        f"class={classes_selected[idx].item()} "
        f"({class_names[classes_selected[idx].item()]}), "
        f"confidence={scores_selected[idx].item():.4f}, "
        f"box=["
        f"{boxes_xyxy[idx,0].item():.1f}, "
        f"{boxes_xyxy[idx,1].item():.1f}, "
        f"{boxes_xyxy[idx,2].item():.1f}, "
        f"{boxes_xyxy[idx,3].item():.1f}]"
    )
```
